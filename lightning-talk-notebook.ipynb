The dataset contains missing values across 11 features, while the remaining 42 features are fully populated with zero missing values. The features with missing values can be categorized into four groups:

Drug Test Metrics: Avg_Days_per_DrugTest (4,260 nulls, ~24%), DrugTests_THC_Positive (3,632 nulls, ~20%), DrugTests_Cocaine_Positive (3,632 nulls, ~20%), DrugTests_Meth_Positive (3,632 nulls, ~20%), and DrugTests_Other_Positive (3,632 nulls, ~20%).

Employment Metrics: Jobs_Per_Year (534 nulls, ~3%) and Percent_Days_Employed (307 nulls, ~2%).

Risk Score: Supervision_Risk_Score_First (330 nulls, ~2%).

Plan for Handling Null Values
Drug Testing Features: The missing values in these columns mean that the individual was never required to take a drug test during their supervision period.

Fill the positive drug test counts with 0 (since no tests were performed).

Fill Avg_Days_per_DrugTest with -1 to indicate the absence of testing.

Add a new binary column DrugTest_Missing (1 if missing, 0 otherwise) to let the model explicitly learn that the individual was not tested.

Categorical & Text Features (Prison_Offense, Gang_Affiliated, Supervision_Level_First):

Fill missing values with the string label "Unknown". This treats unrecorded data as a distinct category without losing any rows.

Employment Features (Percent_Days_Employed, Jobs_Per_Year):

Fill missing values with 0, as unrecorded employment corresponds to zero active employment during supervision.

Risk Score (Supervision_Risk_Score_First):

Impute missing numerical values using the Median of the column since the missing proportion is very small (~2%).

# How many null values does each feature have? what's your plan for each?
### Missing Values & Handling Plan

- **Null Values Count:** The Student Depression dataset typically contains a small number of missing values (primarily in fields like `Financial Stress`, `Academic Pressure`, or `CGPA` depending on the exact version/cleanliness of the raw file).
- **Handling Strategy:**
  - **Categorical Columns (e.g., `Financial Stress`, `Sleep Duration`):** Impute missing entries using the **Mode** (most frequent category) or fill with a generic placeholder like `'Unknown'`.
  - **Numerical Columns (e.g., `CGPA`, `Age`):** Impute missing values using the **Median** to prevent extreme values or outliers from skewing the estimates.
  - **Identifiers / ID:** Drop rows only if the primary `id` is missing or invalid.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Calculate missing values and percentages per column
null_counts = df.isnull().sum()
null_percentages = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame(
    {"Missing Count": null_counts, "Percentage (%)": null_percentages.round(2)}
).sort_values(by="Missing Count", ascending=False)

print("--- Missing Values Report ---")
print(missing_df[missing_df["Missing Count"] > 0])
if missing_df["Missing Count"].sum() == 0:
    print("No missing values found in any column.")

# 2. Automated Imputation Plan Strategy (Code Execution)
df_clean = df.copy()

# Categorical columns imputation (Mode)
cat_cols = df_clean.select_dtypes(
    include=["object", "category", "str"]
).columns
for col in cat_cols:
    if df_clean[col].isnull().sum() > 0:
        mode_val = df_clean[col].mode()[0]
        df_clean[col] = df_clean[col].fillna(mode_val)
        print(f"Imputed missing values in '{col}' using Mode: {mode_val}")

# Numerical columns imputation (Median)
num_cols = df_clean.select_dtypes(include=["number"]).columns
for col in num_cols:
    if df_clean[col].isnull().sum() > 0:
        median_val = df_clean[col].median()
        df_clean[col] = df_clean[col].fillna(median_val)
        print(f"Imputed missing values in '{col}' using Median: {median_val}")
```

### 1. Presence of Outliers
- **`CGPA` & `Age`:** Continuous variables like `CGPA` or `Age` may contain occasional invalid entries or extreme out-of-bounds values (e.g., erroneous `CGPA` entries outside the standard range or improbable age values).
- **Stress & Pressure Scales (`Academic Pressure`, `Work/Study Hours`):** Rating scores (typically on a fixed 1–5 scale) generally do not have extreme mathematical outliers, but `Work/Study Hours` can present extreme values if students report unusually high daily commitments (e.g., >16 hours/day).

---

### 2. Handling Strategy
- **Validation & Boundary Capping:** Verify that bounded metrics (such as `CGPA`) conform to domain constraints. Cap invalid extreme values at realistic domain boundaries (e.g., maximum standard CGPA).
- **Winsorization / IQR Capping:** For extreme tail values in continuous features like `Work/Study Hours`, apply Interquartile Range (IQR) clipping to trim extreme values without losing whole data points.
- **Model Selection Choice:** Utilize tree-based ensemble classifiers (such as Random Forest, XGBoost, or LightGBM), which rely on decision splits rather than distance metrics and are naturally robust against feature outliers.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Define continuous numerical columns for outlier inspection
num_cols = ["Age", "CGPA", "Work/Study Hours"]
num_cols = [col for col in num_cols if col in df.columns]

# 1. Boxplots for visual inspection
fig, axes = plt.subplots(1, len(num_cols), figsize=(5 * len(num_cols), 4))
if len(num_cols) == 1:
    axes = [axes]

for i, col in enumerate(num_cols):
    sns.boxplot(y=df[col], ax=axes[i], color="#3498db")
    axes[i].set_title(f"Outliers in {col}", fontweight="bold")

plt.tight_layout()
plt.show()

# 2. IQR Outlier Detection Function
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers_count = (
        (df[col] < lower_bound) | (df[col] > upper_bound)
    ).sum()
    print(
        f"{col}: Found {outliers_count} outliers (Bounds: [{lower_bound:.2f}, {upper_bound:.2f}])"
    )

# 3. Handle outliers using IQR Capping (Winsorization)
df_clean = df.copy()
for col in num_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Clip values within the upper and lower thresholds
    df_clean[col] = np.clip(df_clean[col], lower_bound, upper_bound)
    
```
